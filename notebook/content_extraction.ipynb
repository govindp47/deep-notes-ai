```python
import re
from pathlib import Path


def clean_bullet_output(text: str) -> str:


    text = text.replace("\r\n", "\n").replace("\r", "\n")

    lines = text.split("\n")

    bullet_pattern = re.compile(
        r'''
        ^\s*
        (?:
            [-*•●○◦▪▸►▶▫]+      # bullet symbols
            |
            \d+[.)]             # numbered bullets
        )
        (?:\s+(.*))?            # optional text on same line
        $
        ''',
        re.VERBOSE,
    )

    separator_pattern = re.compile(
        r'^\s*[-=*_~]{5,}\s*$'
    )

    cleaned_points = []

    current = None

    for raw_line in lines:
        line = raw_line.rstrip()

        # Ignore separator lines
        if separator_pattern.match(line):
            continue

        if not line.strip():
            continue

        match = bullet_pattern.match(line)

        if match:

            if current is not None:
                cleaned_points.append(current.strip())

            text_after_bullet = match.group(1)

            if text_after_bullet:
                current = text_after_bullet.strip()
            else:
                current = ""

        else:

            stripped = line.strip()

            if not stripped:
                continue

            if current is None:
                current = stripped

            elif current == "":
                current = stripped

            else:
                current += " " + stripped

    if current is not None and current.strip():
        cleaned_points.append(current.strip())

    # Collapse multiple spaces
    cleaned_points = [
        re.sub(r"[ \t]+", " ", p).strip()
        for p in cleaned_points
    ]

    output = []

    for i, point in enumerate(cleaned_points, start=1):
        output.append(f"{i}. {point}")

    return "\n".join(output)

```

```python
from __future__ import annotations

from typing import List
from pydantic import BaseModel, Field


class TopicNode(BaseModel):
    """
    Represents one topic or content region in the transcript hierarchy.
    """

    name: str = Field(
        description=(
            'Topic title. '
            'Use "CONTENT" ONLY when this node represents transcript content '
            'that cannot be further grouped into meaningful subtopics.'
        )
    )

    start_point: int = Field(
        description="Inclusive starting transcript point number."
    )

    end_point: int = Field(
        description="Inclusive ending transcript point number."
    )

    children: List["TopicNode"] = Field(
        default_factory=list,
        description="Child topic nodes in the same order as the transcript."
    )


TopicNode.model_rebuild()

class TranscriptHierarchy(BaseModel):
    """
    Represents the complete hierarchical organization of a transcript.

    The hierarchy consists of one or more top-level topic nodes that
    recursively decompose the transcript into progressively smaller
    topics and content regions while preserving the original transcript
    order and point ranges.
    """

    hierarchy: List[TopicNode] = Field(
        description="Top-level hierarchy nodes of the transcript."
    )
```

```python
from __future__ import annotations

import json
import math
import re
from dataclasses import dataclass, asdict, field, asdict, is_dataclass
from pathlib import Path
from typing import Any, Dict, List, Optional, Literal
from uuid import uuid4
from itertools import pairwise


@dataclass(slots=True)
class ContentStoreItem:
    content: str = ""
    summary: str = ""


@dataclass(slots=True)
class ContentNode:
    type: Literal["content"] = "content"
    id: str = ""


@dataclass(slots=True)
class TitleNode:
    type: Literal["topic"] = "topic"
    name: str = ""
    subtopics: list["Node"] = field(default_factory=list)


Node = TitleNode | ContentNode

@dataclass(slots=True)
class ContentExtraction:
    id: str
    hierarchy_path: list[str]
    starting_point: int
    ending_point: int


@dataclass(slots=True)
class ExtractionResult:
    extracted: list[ContentExtraction]
    metadata: dict[str, ContentStoreItem]
    node: Node


@dataclass(slots=True)
class ContentPayload:
    id: str
    hierarchy_path: list[str]
    range: tuple[int, int]
    transcript_points: list[str]


@dataclass(slots=True)
class StructuredContentPayload:
    id: str
    hierarchy_path: list[str]
    structured_content: str


@dataclass(slots=True)
class PayloadResult:
    payload: list[ContentPayload]
    metadata: dict[str, ContentStoreItem]
    nodes: list[Node]


def json_serializer(obj):
    if is_dataclass(obj):
        return asdict(obj)
    raise TypeError(f"Object of type {type(obj).__name__} is not JSON serializable")


def load_json(path: str | Path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def save_json(obj, path: str | Path):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(obj, f, indent=4, ensure_ascii=False)


def load_numbered_points(file: str | Path) -> List[str]:
    """
    Loads numbered transcript points from a txt file.

    Returns:
        List[str]
            index 0 -> point 1
            index 1 -> point 2
            ...

    Raises:
        AssertionError:
            If the numbering is not continuous.
    """

    POINT_PATTERN = re.compile(r"^\s*(\d+)\.\s+(.*)$")

    points: List[str] = []
    point_numbers: List[int] = []

    with open(file, encoding="utf-8") as f:
        for line in f:
            m = POINT_PATTERN.match(line.rstrip())
            if m:
                point_numbers.append(int(m.group(1)))
                points.append(line.strip())

    if points:
        # Ensure numbering is exactly 1..N
        expected = list(range(1, len(points) + 1))
        assert point_numbers == expected, (
            "Invalid numbering detected.\n"
            f"Expected: {expected[:10]}...{expected[-10:]}\n"
            f"Found:    {point_numbers[:10]}...{point_numbers[-10:]}"
        )

        # Additional sanity check
        assert len(points) == point_numbers[-1], (
            f"Expected {len(points)} points but last point number is "
            f"{point_numbers[-1]}."
        )

    return points


def equal_partition_last_points(file: str | Path, n: int) -> List[int]:
    """
    Splits the transcript into approximately equal character-length parts
    and returns the last point number in each part.

    Example:
        Transcript:
            1. ...
            2. ...
            ...
            100. ...

        n = 4

        Returns something like:
            [27, 52, 76, 100]

    The partitions are based on transcript length, not number of points.
    """

    if n <= 0:
        raise ValueError("n must be greater than zero.")
    
    transcript = file.read_text(encoding="utf-8")
    
    POINT_PATTERN = re.compile(r"^(\d+)\.\s", re.MULTILINE)

    matches = list(POINT_PATTERN.finditer(transcript))

    if not matches:
        return []

    boundaries = []
    transcript_length = len(transcript)

    for part in range(1, n + 1):

        target_position = math.ceil(transcript_length * part / n)
        last_point = int(matches[-1].group(1))

        for i, match in enumerate(matches):
            if match.start() > target_position:
                last_point = int(matches[i - 1].group(1))
                break

        boundaries.append(last_point)

    # Ensure final partition ends at the final point
    boundaries[-1] = int(matches[-1].group(1))

    return boundaries


def build_partition_ranges(last_points: list[int]) -> list[tuple[int, int]]:
    """
    [145, 287, 421] -> [(1,145), (146,287), (288,421)]
    """

    if not last_points:
        return []

    ranges = [(1, last_points[0])]

    for previous, current in pairwise(last_points):
        ranges.append((previous + 1, current))

    return ranges



def _extract_content_nodes(
    node: TopicNode,
    path: tuple[str, ...] = (),
) -> ExtractionResult:
    """
    Traverses the TopicNode hierarchy once.

    Simultaneously:

    1. Extracts every CONTENT node.
    2. Generates UUIDs.
    3. Creates the metadata dictionary.
    4. Builds the lightweight Node hierarchy.
    """

    if node.name == "CONTENT":
        content_id = str(uuid4())

        return ExtractionResult(
            extracted=[
                ContentExtraction(
                    id=content_id,
                    hierarchy_path=list(path),
                    starting_point=node.start_point,
                    ending_point=node.end_point,
                )
            ],
            metadata={
                content_id: ContentStoreItem(),
            },
            node=ContentNode(
                id=content_id,
            ),
        )

    current_path = (*path, node.name)
    extracted: list[ContentExtraction] = []
    metadata: dict[str, ContentStoreItem] = {}
    children: list[Node] = []

    for child in node.children:

        result = _extract_content_nodes(
            child,
            current_path,
        )

        extracted.extend(result.extracted)
        metadata.update(result.metadata)
        children.append(result.node)

    return ExtractionResult(
        extracted=extracted,
        metadata=metadata,
        node=TitleNode(
            name=node.name,
            subtopics=children,
        ),
    )



def build_content_payloads(
    hierarchy: list[TopicNode],
    transcript_points: list[str],
) -> PayloadResult:
    """
    Returns

    payload:
        {
            content_uuid: ContentPayload(...)
        }

    metadata:
        {
            content_uuid: ContentStoreItem(...)
        }
    """

    all_extracted: list[ContentExtraction] = []
    metadata: dict[str, ContentStoreItem] = {}
    converted_hierarchy: list[Node] = []

    for topic in hierarchy:
        result = _extract_content_nodes(topic)

        all_extracted.extend(result.extracted)
        metadata.update(result.metadata)
        converted_hierarchy.append(result.node)

    previous_end = 0
    payload: list[ContentPayload] = []

    for node in all_extracted:

        start = node.starting_point
        end = node.ending_point

        # Fill uncovered gaps.
        if previous_end + 1 < start:
            start = previous_end + 1

        previous_end = max(previous_end, end)

        payload.append(
            ContentPayload(
                id=node.id,
                hierarchy_path=node.hierarchy_path,
                range=(start, end),
                transcript_points=transcript_points[start - 1 : end],
            )
        )

    return PayloadResult(
        payload=payload,
        metadata=metadata,
        nodes=converted_hierarchy,
    )


def filter_payload_by_range(
    payload: list[ContentPayload],
    start_point: int,
    end_point: int,
) -> list[ContentPayload]:
    """
    Returns only the CONTENT nodes whose ending point lies within the
    specified filter range.
    """

    return [
        item
        for item in payload
        if start_point <= item.range[1] <= end_point
    ]
```

```python
class StructuredContent(BaseModel):
    """
    Structured markdown generated for a single CONTENT node.
    """

    id: str = Field(
        description=(
            "Temporary CONTENT node identifier "
            "(for example N1, N2, N3) exactly as provided in the input."
        )
    )

    markdown: str = Field(
        description=(
            "Markdown representation of ONLY the transcript points belonging "
            "to this CONTENT node."
        )
    )


class StructuredContentBatch(BaseModel):
    """
    Collection of structured markdown outputs for all CONTENT nodes processed
    in a single LLM request. Each item corresponds to exactly one CONTENT node
    from the input and must preserve the same UUID.
    """

    items: list[StructuredContent]


class ContentSummary(BaseModel):
    """
    Summary generated for a single CONTENT node.
    """

    id: str = Field(
        description=(
            "Temporary CONTENT node identifier "
            "(for example N1, N2, N3) exactly as provided in the input."
        )
    )

    summary: str = Field(
        description=(
            "Markdown summary of ONLY the transcript content belonging "
            "to this CONTENT node."
        )
    )


class ContentSummaryBatch(BaseModel):
    """
    Collection of summaries for all CONTENT nodes processed in a single LLM
    request. Each item corresponds to exactly one CONTENT node from the input
    and preserves the same UUID.
    """

    items: list[ContentSummary]

def save_nodes_hierarchy(
    updated_nodes_hierarchy: list[Node],
    output_file: str | Path,
) -> None:
    """
    Save nodes_hierarchy (list[Node]) as JSON.
    """
    
    with output_file.open("w", encoding="utf-8") as f:
        json.dump(
            updated_nodes_hierarchy,
            f,
            default=json_serializer,
            indent=4,
            ensure_ascii=False,
        )


def save_nodes_content(
    nodes_content: dict[str, ContentStoreItem],
    output_file: str | Path,
) -> None:
    """
    Save nodes_content (dict[str, ContentStoreItem]) as JSON.
    """

    with output_file.open("w", encoding="utf-8") as f:
        json.dump(
            nodes_content,
            f,
            default=json_serializer,
            indent=4,
            ensure_ascii=False,
        )


def validate_batch_response(
    response: Any,
    expected_ids: set[str],
    entity_name: str = "CONTENT node",
) -> None:
    """
    Validates an LLM structured-output batch.

    Checks:
    - correct number of items
    - no duplicate ids
    - returned ids exactly match expected ids
    """

    items = response.items
    returned_ids = [item.id for item in items]

    if len(items) != len(expected_ids):
        raise ValueError(
            f"Expected {len(expected_ids)} {entity_name}s, "
            f"but LLM returned {len(items)}."
        )

    duplicates = {
        node_id
        for node_id in returned_ids
        if returned_ids.count(node_id) > 1
    }

    if duplicates:
        raise ValueError(
            f"Duplicate {entity_name} ids returned by LLM: "
            f"{sorted(duplicates)}"
        )

    returned_id_set = set(returned_ids)

    if returned_id_set != expected_ids:
        missing = sorted(expected_ids - returned_id_set)
        unexpected = sorted(returned_id_set - expected_ids)

        raise ValueError(
            f"LLM returned incorrect {entity_name} ids.\n"
            f"Missing: {missing}\n"
            f"Unexpected: {unexpected}"
        )
```

```python
MAX_RETRIES = 2
INITIAL_PARTITIONS = 4
FALLBACK_PARTITIONS = 6

structured_llm = ChatOpenAI(
    model="gpt-5-mini",  # or "gpt-5.4-mini" or "gpt-4o-mini"
    temperature=0
).with_structured_output(StructuredContentBatch)

prompt = ChatPromptTemplate.from_template(content_nodes_prompt)

chain = prompt | structured_llm

def invoke_structured_batch(
    filtered_payload: list[ContentPayload],
    retry_count: int = MAX_RETRIES,
) -> StructuredContentBatch:
    """
    Calls the LLM and retries if the response contains duplicate
    or incorrect ids.

    Raises:
        ValueError:
            If the response length does not match.
        RuntimeError:
            If retries are exhausted.
    """

    expected_ids = {
        item.id
        for item in filtered_payload
    }

    last_error: Exception | None = None

    for attempt in range(1, retry_count + 1):

        response: StructuredContentBatch = chain.invoke(
            {
                "NODES_CONTENT": json.dumps(
                    filtered_payload,
                    default=json_serializer,
                    indent=2,
                    ensure_ascii=False,
                )
            }
        )

        try:
            validate_batch_response(
                response=response,
                expected_ids=expected_ids,
            )

            return response

        except ValueError as e:

            last_error = e
            message = str(e)
            print(message)

            # Wrong number of items.
            # Caller will repartition.
            if "Expected" in message and "but LLM returned" in message:
                raise

            print(
                f"Retry {attempt}/{retry_count} "
                "because ids were invalid..."
            )

    raise RuntimeError(
        "Failed to generate valid structured output "
        f"after {retry_count} retries."
    ) from last_error


def _process_partitions(
    partition_count: int,
) -> None:
    """
    Processes all transcript partitions.
    """

    partition_points = equal_partition_last_points(
        transcript_file,
        partition_count,
    )

    partition_ranges = build_partition_ranges(
        partition_points,
    )

    for start_point, end_point in partition_ranges:

        print(f"Processing points {start_point} -> {end_point}")

        filtered_payload = filter_payload_by_range(
            payload=payload,
            start_point=start_point,
            end_point=end_point,
        )

        if not filtered_payload:
            continue

        temp_payload = []
        temp_to_real = {}

        for index, item in enumerate(filtered_payload, start=1):
            temp_id = f"N{index}"
            temp_to_real[temp_id] = item.id

            temp_payload.append(
                ContentPayload(
                    id=temp_id,
                    hierarchy_path=item.hierarchy_path,
                    range=item.range,
                    transcript_points=item.transcript_points,
                )
            )

        response = invoke_structured_batch(temp_payload)

        for item in response.items:
            real_id = temp_to_real[item.id]
            nodes_content[real_id].content = item.markdown


def generate_structured_content() -> None:
    """
    Generates structured markdown for every CONTENT node.

    Strategy:

    1. Try using INITIAL_PARTITIONS.
    2. If any partition returns fewer/more items than expected,
       repartition into FALLBACK_PARTITIONS and retry once.
    3. Duplicate ids / incorrect ids are automatically retried by
       invoke_structured_batch().
    """

    try:

        _process_partitions(INITIAL_PARTITIONS)

    except ValueError as e:

        message = str(e)
        print(message)

        if (
            "Expected" not in message
            or "but LLM returned" not in message
        ):
            raise

        print(
            "\nLLM failed to return all CONTENT nodes.\n"
            f"Retrying with {FALLBACK_PARTITIONS} partitions...\n"
        )

        _process_partitions(FALLBACK_PARTITIONS)
```

```python
MAX_RETRIES = 2
INITIAL_SUMMARY_PARTITIONS = 4
FALLBACK_SUMMARY_PARTITIONS = 6

node_summary_llm = ChatOpenAI(
    model="gpt-4o-mini",  # or "gpt-5.4-mini" or "gpt-4o-mini"
    temperature=0
).with_structured_output(ContentSummaryBatch)

summary_prompt = ChatPromptTemplate.from_template(nodes_summary_prompt)

summary_chain = summary_prompt | node_summary_llm

def invoke_summary_batch(
    summary_input: list[StructuredContentPayload],
    retry_count: int = MAX_RETRIES,
) -> ContentSummaryBatch:
    """
    Calls the summary LLM.

    Retries only when duplicate/incorrect ids are returned.

    Raises:
        ValueError:
            If response length does not match.
        RuntimeError:
            If retries are exhausted.
    """

    expected_ids = {
        item.id
        for item in summary_input
    }

    last_error: Exception | None = None

    for attempt in range(1, retry_count + 1):

        response: ContentSummaryBatch = summary_chain.invoke(
            {
                "NODES_CONTENT": json.dumps(
                    summary_input,
                    default=json_serializer,
                    indent=2,
                    ensure_ascii=False,
                )
            }
        )

        try:
            validate_batch_response(
                response=response,
                expected_ids=expected_ids,
            )

            return response

        except ValueError as e:

            last_error = e
            message = str(e)
            print(message)

            # Length mismatch -> caller will repartition.
            if "Expected" in message and "but LLM returned" in message:
                raise

            print(
                f"Retry {attempt}/{retry_count} "
                "because summary ids were invalid..."
            )

    raise RuntimeError(
        f"Failed to generate valid summaries after {retry_count} retries."
    ) from last_error


def _process_summary_partitions(
    partition_count: int,
) -> None:
    """
    Processes summary generation for all partitions.
    """

    partition_points = equal_partition_last_points(
        transcript_file,
        partition_count,
    )

    partition_ranges = build_partition_ranges(
        partition_points,
    )

    for start_point, end_point in partition_ranges:

        print(f"Processing summary points {start_point} -> {end_point}")

        filtered_payload = filter_payload_by_range(
            payload=payload,
            start_point=start_point,
            end_point=end_point,
        )

        if not filtered_payload:
            continue

        summary_input: list[StructuredContentPayload] = []
        temp_to_real: dict[str, str] = {}

        for index, payload_item in enumerate(filtered_payload, start=1):
            temp_id = f"N{index}"
            temp_to_real[temp_id] = payload_item.id

            summary_input.append(
                StructuredContentPayload(
                    id=temp_id,
                    hierarchy_path=payload_item.hierarchy_path,
                    structured_content=nodes_content[payload_item.id].content,
                )
            )

        response = invoke_summary_batch(summary_input)

        for item in response.items:
            real_id = temp_to_real[item.id]
            nodes_content[real_id].summary = item.summary


def generate_summaries() -> None:
    """
    Generates summaries for every CONTENT node.

    Strategy:
    1. Try using INITIAL_SUMMARY_PARTITIONS.
    2. If any partition returns fewer/more items than expected,
       retry once using FALLBACK_SUMMARY_PARTITIONS.
    3. Duplicate/incorrect ids are automatically retried by
       invoke_summary_batch().
    """

    try:

        _process_summary_partitions(
            INITIAL_SUMMARY_PARTITIONS,
        )

    except ValueError as e:

        message = str(e)
        print(message)

        if (
            "Expected" not in message
            or "but LLM returned" not in message
        ):
            raise

        print(
            "\nSummary generation returned an incorrect number of items.\n"
            f"Retrying with {FALLBACK_SUMMARY_PARTITIONS} partitions...\n"
        )

        _process_summary_partitions(
            FALLBACK_SUMMARY_PARTITIONS,
        )
```

```python
TRANSCRIPT_TITLE = "LangGraph Course"

HIERARCHY_FILE = Path("nodes_hierarchy.json")
CONTENT_FILE = Path("nodes_content2.json")

CONTENT_OUTPUT = Path("course_content.md")
SUMMARY_OUTPUT = Path("course_summary.md")


def _load_node(data: dict) -> Node:
    """
    Recursively reconstruct Node dataclasses.
    """

    node_type = data["type"]

    if node_type == "content":
        return ContentNode(
            id=data["id"],
        )

    if node_type == "topic":
        return TitleNode(
            name=data["name"],
            subtopics=[
                _load_node(child)
                for child in data.get("subtopics", [])
            ],
        )

    raise ValueError(f"Unknown node type: {node_type}")


def load_nodes_hierarchy(
    path: str | Path,
) -> list[Node]:
    """
    Reverse of save_nodes_hierarchy().
    """

    raw = load_json(path)

    return [
        _load_node(node)
        for node in raw
    ]


def load_nodes_content(
    path: str | Path,
) -> dict[str, ContentStoreItem]:
    """
    Reverse of save_nodes_content().
    """

    raw = load_json(path)

    return {
        node_id: ContentStoreItem(
            content=value["content"],
            summary=value["summary"],
        )
        for node_id, value in raw.items()
    }


# ============================================================================
# MARKDOWN RENDERER
# ============================================================================

def _render_node(
    node: Node,
    content_store: dict[str, ContentStoreItem],
    output: list[str],
    *,
    summary: bool,
    heading_level: int,
):
    """
    Recursively renders a Node hierarchy.
    """

    if isinstance(node, TitleNode):

        output.append(f'{"#" * heading_level} {node.name}')
        output.append("")

        for child in node.subtopics:
            _render_node(
                child,
                content_store,
                output,
                summary=summary,
                heading_level=heading_level + 1,
            )

        return

    # Content node

    item = content_store[node.id]
    text = item.summary if summary else item.content
    text = text.strip()

    if text:
        output.append(text)
        output.append("")


def build_markdown_document(
    *,
    transcript_title: str,
    hierarchy: list[Node],
    content_store: dict[str, ContentStoreItem],
    summary: bool = False,
) -> str:
    """
    Creates the complete markdown document.
    """

    output: list[str] = [
        f"# {transcript_title}",
        "",
    ]

    for node in hierarchy:
        _render_node(
            node,
            content_store,
            output,
            summary=summary,
            heading_level=2,
        )

    return "\n".join(output).rstrip() + "\n"



def save_markdown(
    markdown: str,
    path: str | Path,
):
    Path(path).write_text(
        markdown,
        encoding="utf-8",
    )
```

```python
hierarchy = load_nodes_hierarchy(HIERARCHY_FILE)
nodes_content = load_nodes_content(CONTENT_FILE)

content_md = build_markdown_document(
    transcript_title=TRANSCRIPT_TITLE,
    hierarchy=hierarchy,
    content_store=nodes_content,
    summary=False,
)

summary_md = build_markdown_document(
    transcript_title=TRANSCRIPT_TITLE,
    hierarchy=hierarchy,
    content_store=nodes_content,
    summary=True,
)

save_markdown(content_md, CONTENT_OUTPUT)
save_markdown(summary_md, SUMMARY_OUTPUT)

print(f"Saved: {CONTENT_OUTPUT}")
print(f"Saved: {SUMMARY_OUTPUT}")
```
